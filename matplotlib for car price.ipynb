```python
import pandas as pd
df = pd.read_csv("C:\\Users\\anilr\\Downloads\\Used Car Price & Market Analysis Project.xls")
df.head()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

sns.histplot(df['Selling_Price'], kde=True)
plt.title('Distribution of Selling Price')
plt.show()
```

```python
sns.countplot(data=df, 
              x='Fuel_Type'
             )
plt.title('Number of Cars by Fuel Type')
plt.show()
```

```python
sns.barplot(data=df,
                x='Vehicle_Age',
                y='Selling_Price')
plt.show()
```

```python
sns.scatterplot(data=df,
                x='Km_Driven',
                y='Selling_Price')
plt.show()
```

```python
sns.barplot(data=df,
            x='Fuel_Type',
            y='Selling_Price')
plt.show()
```

```python
sns.scatterplot(
    data=df,
    x='Km_Driven',
    y='Selling_Price',
    hue='Fuel_Type',
    style='Transmission_Type'
)
plt.show()
```

```python
sns.pairplot(
    df[['Vehicle_Age', 'Km_Driven', 'Mileage',
        'Engine', 'Max_Power', 'Selling_Price']]
)
plt.show()
```

```python
plt.figure(figsize=(10, 6))

sns.heatmap(
    df.select_dtypes(include='number').corr(),
    annot=True
   
)

plt.title('Correlation Heatmap')
plt.show()
```

```python
sns.boxplot(
    y=df['Selling_Price']
)
plt.show()
```

```python
sns.boxplot(
    y=df['Km_Driven']
)
plt.show()
```

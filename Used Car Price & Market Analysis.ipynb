# Cleaning tasks:

```python
numeric_cols = ['Km_Driven', 'Mileage', 'Engine', 
                'Max_Power', 'Seats', 'Selling_Price']

for col in numeric_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    outliers = df[(df[col] < lower_limit) | (df[col] > upper_limit)]

    print(col, "Outliers:", len(outliers))
```

```python
categorical_cols = ['Car_Name', 'Brand', 'Model',
                    'Seller_Type', 'Fuel_Type',
                    'Transmission_Type']

for col in categorical_cols:
    df[col] = df[col].str.strip()
df[col].head()
```

```python
print(df[numeric_cols].describe())
```

```python
print("Negative Km:", (df['Km_Driven'] < 0).sum())
print("Negative Price:", (df['Selling_Price'] < 0).sum())
print("Invalid Mileage:", (df['Mileage'] <= 0).sum())
print("Invalid Engine:", (df['Engine'] <= 0).sum())
```
